# BRCA

```python
ad = sc.read_h5ad('/depot/natallah/data/Luopin/Metastasis_single_cell/Data/Cancer_cell_data_reprocessed/BRCA_integrated.harmony.h5ad')
ad
```

```python
# init
ad.layers["counts"] = ad.raw.X.copy()
scvi.model.SCVI.setup_anndata(ad, layer="counts", batch_key="Project_ID")
```

```python
# 2. Create and train the model
model = scvi.model.SCVI(ad, n_latent=30)  # latent dimension, typically 10-30
model.train()
```

```python
SCVI_LATENT_KEY = "X_scVI"
ad.obsm[SCVI_LATENT_KEY] = model.get_latent_representation()
sc.pp.neighbors(ad, use_rep=SCVI_LATENT_KEY)
sc.tl.leiden(ad)
sc.tl.umap(ad)
```

```python
sc.pl.umap(
    ad,
    color=["Project_ID"],
)
```

```python
ad.write_h5ad('/depot/natallah/data/Luopin/Metastasis_single_cell/Data/scVI/BRCA_integrated.scVI.h5ad', compression='gzip')
```
